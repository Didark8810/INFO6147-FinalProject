# 🧠 GradCAM con EfficientNet (ejemplo en `uno/f7.jpg`)

```
import torch
import torch.nn.functional as F
from torchvision import models, transforms
from PIL import Image
import matplotlib.pyplot as plt
import numpy as np
import cv2
import os
import io
```

```
transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])
])

image_path = "uno/f7.jpg"
image = Image.open(image_path).convert("RGB")
input_tensor = transform(image).unsqueeze(0)
```

```
def generate_gradcam(model, input_tensor, target_layer="features", class_idx=None):
    model.eval()
    
    activations = []
    gradients = []

    def forward_hook(module, input, output):
        activations.append(output.detach())

    def backward_hook(module, grad_input, grad_output):
        gradients.append(grad_output[0].detach())

    target_module = dict([*model.named_modules()])[target_layer]
    forward_handle = target_module.register_forward_hook(forward_hook)
    backward_handle = target_module.register_backward_hook(backward_hook)

    output = model(input_tensor)
    if class_idx is None:
        class_idx = output.argmax().item()

    score = output[:, class_idx]
    score.backward()

    act = activations[0][0]
    grad = gradients[0][0]

    weights = grad.mean(dim=(1, 2))
    cam = torch.zeros(act.shape[1:], dtype=torch.float32)

    for i, w in enumerate(weights):
        cam += w * act[i]

    cam = F.relu(cam)
    cam -= cam.min()
    cam /= cam.max()
    cam = cam.cpu().numpy()
    cam = cv2.resize(cam, (224, 224))

    forward_handle.remove()
    backward_handle.remove()
    return cam
```

```
from models.classifier import FruitClassifier

model_path = "models_trained/efficientnet_finetuned.pth"
label_path = "models_trained/class_labels.txt"

with open(label_path, 'r') as f:
    class_names = [line.strip() for line in f.readlines()]

model = FruitClassifier(num_classes=len(class_names))
model.load_model(model_path)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model.to(device)
input_tensor = input_tensor.to(device)
```

```
with torch.no_grad():
    output = model(input_tensor)
    probs = F.softmax(output[0], dim=0)
    pred_idx = torch.argmax(probs).item()
    pred_label = class_names[pred_idx]
    confidence = probs[pred_idx].item()

heatmap_np = generate_gradcam(model, input_tensor, target_layer="features", class_idx=pred_idx)

img_np = np.array(image.resize((224, 224)))
heatmap_color = cv2.applyColorMap(np.uint8(255 * heatmap_np), cv2.COLORMAP_JET)
heatmap_color = cv2.cvtColor(heatmap_color, cv2.COLOR_BGR2RGB)
overlay = cv2.addWeighted(img_np, 0.5, heatmap_color, 0.5, 0)

plt.figure(figsize=(12, 5))
plt.subplot(1, 3, 1)
plt.imshow(img_np)
plt.title("Imagen original")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(heatmap_np, cmap='jet')
plt.title("Mapa de Calor (GradCAM)")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(overlay)
plt.title(f"Superpuesto: {pred_label} ({confidence:.2f})")
plt.axis("off")
plt.tight_layout()
plt.show()
```
